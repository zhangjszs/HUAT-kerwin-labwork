# 第二题
环境创建写为了一个bash
```bash
#!/bin/bash

# Step 1: 将依赖放入 Kettle 的 lib 中
echo "Copying MySQL connector JAR to Kettle's lib directory..."
cp /data/bigfiles/mysql-connector-java-5.1.41-bin.jar /opt/kettle/data-integration/lib

# Step 2: 创建 MySQL 环境
echo "Starting MySQL service..."
service mysql start

echo "Creating database and table in MySQL..."
mysql -u root -p123123 << EOF
CREATE DATABASE IF NOT EXISTS kettle;
USE kettle;
DROP TABLE IF EXISTS book;
CREATE TABLE book(
    no int,
    author VARCHAR(10),
    price int,
    amount int
) ENGINE=InnoDB DEFAULT CHARSET=utf8;
EOF

# Step 3: 启动 Kettle
echo "Starting Kettle..."
/opt/kettle/data-integration/spoon.sh
```

# 第三题
环境创建写为了一个bash
```bash
#!/bin/bash

# Step 1: 将依赖放入 Kettle 的 lib 中，并启动 Kettle
echo "Copying MySQL connector JAR to Kettle's lib directory..."
cp /data/bigfiles/mysql-connector-java-5.1.41-bin.jar /opt/kettle/data-integration/lib

echo "Starting Kettle..."
/opt/kettle/data-integration/spoon.sh &

# Step 2: 创建 MySQL 环境
echo "Starting MySQL service..."
service mysql start

echo "Creating database and table in MySQL..."
mysql -u root -p123123 << EOF
CREATE DATABASE IF NOT EXISTS kettle;
USE kettle;
DROP TABLE IF EXISTS student_info;
CREATE TABLE student_info(
    sno int,
    sname VARCHAR(10),
    ssex VARCHAR(2),
    sage int
) ENGINE=InnoDB DEFAULT CHARSET=utf8;
EOF

echo "Setup complete."
```
